Создать класс Detector со следующими параметрами:
- R = 2 метра

Фотоумножители расположить по поверхности детектора достаточно равномерно. Координаты фотоумножителей сохранить в наиболее подходящую для этого структуру. Разработать метод, определяющий, принадлежит ли точка на поверхности детектора какому-либо фотоумножителю. Визуализировать геометрию детектора с помощью пакета k3d (опционально).

Для решения домашнего задания я использовала материал этой статьи на хабре: https://habr.com/ru/articles/506526/

Итого:
- Собственно, для равномерного расположения фотоумножителей по поверхности использовался алгоритм Фибоначчи
- Координаты сохраняются в массив, потому что я не уверена, куда ещё их можно запихнуть
- Вроде метод работает

```python
import math
import numpy as np
from numpy import arange, pi, sin, cos, arccos
```

```python
class Detector:
    
    def __init__(self, R=2.0, N=40, r_cm=10.0, a=0.5):
        self.R = float(R)
        self.N = int(N)
        self.r = float(r_cm) / 100.0
        self.a = float(a)
        self.centers = np.zeros((self.N, 3))
        self.distribute_on_detector()

    def distribute_on_detector(self):
        golden_ratio = (1 + 5 ** 0.5) / 2
        
        for i in range(self.N):
            theta = 2 * pi * i / golden_ratio
            phi = arccos(1 - 2 * (i + 0.5) / self.N)
            x, y, z = self.R * cos(theta) * sin(phi), self.R * sin(theta) * sin(phi), self.R * cos(phi)
            self.centers[i] = [x, y, z]

    def is_point_on(self, point):
        point = np.array(point)
        distances = np.linalg.norm(self.centers - point, axis=1)
        result = bool(np.any(distances <= self.r))
        return result
```

```python
detector = Detector(R=2, N=40, r_cm=10, a=0.5)
test_point = detector.centers[0] 
print(f"Точка в центре ФЭУ №0 принадлежит ФЭУ? {detector.is_point_on(test_point)}")
```
